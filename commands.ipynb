```python


models_to_try = [
      "NCSF", 
    #  "NSF", 
    # "SOSPF", 
  #  "CNF",
   # "UNAF"
]



collections = {
  "OTBC": "OuterTrackerBarrelCollection",
  "ITBC": "InnerTrackerBarrelCollection",
  "VBC": "VertexBarrelCollection",
     "OTEC": "OuterTrackerEndcapCollection",
   "ITEC": "InnerTrackerEndcapCollection",
   "VEC": "VertexEndcapCollection",
    

}
    
model_sizes = {
    # "S": [3, "128,128,128"],
      "M": [5, "128,128,128"],
     #"L": [3, "256,256,256"],
    # "XL": [5, "256,256,256"]
    #"XXL": [7, "128,128,128"],
}

model_sizes_CNF = {
  #  "S": [3, "128,128,128"],
    "S": [5, "256,256,256,256"],
     "M": [8, "256,256,256,256"],
    "L": [3, "512,512,512"],
    "XL": [5, "512,512,512"]
    

}




name = "cond5"

for m in models_to_try:
    for ms in model_sizes.keys():
        if m != "CNF":
            t = model_sizes[ms][0]
            hf = model_sizes[ms][1]
            num_epochs = 1000
            batch_size = 2048
        else:
            t = model_sizes_CNF[ms][0]
            hf = model_sizes_CNF[ms][1]
            num_epochs = 200
            batch_size = 8192
        for col in collections.keys():

           
            
            col_name = collections[col]

            # default feature order: E, r, phi, z, t, side, layer
            

            if "Barrel" in col_name:
                feature_order = "0,4,1,2,3,5,6,7,8"  # phi E t r, z side layer; condition on side, layer
                n_cond_inputs = 4
            elif "Endcap" in col_name: 
                feature_order = "0,4,1,2,3,5,6,7,8"  # phi E t r z side layer; condition on z, side, layer
                n_cond_inputs = 4

                
            command = f"python run_zuko.py \
            --ZUKO_ID {m}\
            --NUM_EPOCHS {num_epochs}\
            --TRANSFORMS {t} \
            --FREQS {t} \
            --HIDDEN_FEATURES {hf} \
            --COLLECTION_LIST {col_name} \
            --TRAIN_FLOW \
            --EVAL_FLOW \
            --NAME {col}_m3p3_cond{n_cond_inputs}\
            --BATCH_SIZE {batch_size} \
           --LEARNING_RATE 0.0001 \
            --NUM_COND_INPUTS {n_cond_inputs}\
            --FEATURE_ORDER {feature_order}\
            --BINS 48\
            --DEGREE 4\
            --POLYNOMIALS 5 \
            --FEATURES rphi"
            print(command)
                
```
